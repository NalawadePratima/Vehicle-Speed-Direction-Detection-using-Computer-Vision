## Approach
**Speed & direction from homography projection (more realistic, bird’s-eye view).**


```python
import cv2
import numpy as np
import math
from ultralytics import YOLO
from supervision import ByteTrack, Detections, BoxAnnotator, LabelAnnotator

# ----------------------------
# Load trained YOLO model
# ----------------------------
model = YOLO("/content/best.pt")

# Tracker + annotators
tracker = ByteTrack()
box_annotator = BoxAnnotator()
label_annotator = LabelAnnotator()

# ----------------------------
# Video input/output
# ----------------------------
video_path = "/content/Input_video.mp4"
cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
out = cv2.VideoWriter(
    "output_withspeed_dir_tail.mp4",
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps,
    (int(cap.get(cv2.CAP_PROP_FRAME_WIDTH)),
     int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT)))
)

# ----------------------------
# Homography (from calibration)
# ----------------------------
H = np.array([
    [-2.91650089e-02, -6.14513566e-02,  4.89978978e+01],
    [ 7.46458798e-04,  2.85396080e-01, -2.00214183e+02],
    [ 5.18942498e-05, -3.43950385e-03,  1.00000000e+00]
])

def project_point(x, y, H):
    pt = np.array([[[x, y]]], dtype=np.float32)
    dst = cv2.perspectiveTransform(pt, H)
    return dst[0][0]  # (X, Y) in meters

# ----------------------------
# Tracking storage
# ----------------------------
last_positions = {}      # tracker_id → (X, Y)
trajectories = {}        # tracker_id → list of (cx, cy) for tails


# ----------------------------
# Processing loop
# ----------------------------
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame, verbose=False)[0]
    detections = Detections.from_ultralytics(results)
    detections = tracker.update_with_detections(detections)

    labels = []
    for xyxy, class_id, tracker_id, conf in zip(
            detections.xyxy, detections.class_id,
            detections.tracker_id, detections.confidence):

        if tracker_id is None:
            continue

        # pixel center
        x1, y1, x2, y2 = xyxy
        cx, cy = int((x1 + x2) / 2), int((y1 + y2) / 2)

        # project to meters
        X, Y = project_point(cx, cy, H)

        # calculate speed + direction
        speed_kmh = 0.0
        #direction = ""
        if tracker_id in last_positions:
            prevX, prevY = last_positions[tracker_id]
            dist_m = np.sqrt((X - prevX) ** 2 + (Y - prevY) ** 2)
            speed_mps = dist_m * fps
            speed_kmh = speed_mps * 3.6

            dx, dy = X - prevX, Y - prevY
            if abs(dx) < 0.001 and abs(dy) < 0.001:
                direction = ""
            else:
                angle = math.degrees(math.atan2(dy, dx))
                if -45 <= angle <= 45:
                    direction = "right"
                elif 135 <= angle or angle <= -135:
                    direction = "left"
                elif 45 < angle < 135:
                    direction = "up"
                else:
                    direction = "down"

                direction += f"({angle:.1f})"

        last_positions[tracker_id] = (X, Y)

        # update tail history
        if tracker_id not in trajectories:
            trajectories[tracker_id] = []
        trajectories[tracker_id].append((cx, cy))
        if len(trajectories[tracker_id]) > 15:
            trajectories[tracker_id].pop(0)

        # safe label formatting
        label = (
            f"ID {tracker_id} {model.model.names[class_id]}"
            f"{conf*100:.1f}% {speed_kmh:.1f} km/h"
        )
        if direction:
            label += f" {direction}"
        labels.append(label)

    # draw boxes + labels
    frame = box_annotator.annotate(scene=frame, detections=detections)
    frame = label_annotator.annotate(scene=frame, detections=detections, labels=labels)

    # cleanup old trajectories
    active_ids = set(detections.tracker_id)
    trajectories = {tid: pts for tid, pts in trajectories.items() if tid in active_ids}

    # draw tails
    for tr_id, pts in trajectories.items():
        for j in range(1, len(pts)):
            cv2.line(frame, pts[j - 1], pts[j], (0, 255, 255), 2)

    out.write(frame)

cap.release()
out.release()
print("Video saved as output_withspeed_direction_tail.mp4")
```
